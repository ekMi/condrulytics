# Durée

```python
import re

pattern = re.compile(
    r"^(\d{1,2}:\d{2}:\d{2}|\d{2}:\d{2}|\d{2}:\d{2}:\d{1})$"
)

invalides = df.loc[
    ~df["Temps"].astype(str).str.match(pattern, na=False),
    "Temps"
].unique()

print(invalides)
```

```python
df[df["Temps"] == "#VALEUR!"]
```

```python
INVALID_VALUES = {
    "#VALEUR!"
}

def normalize_time(value):

    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # valeurs invalides connues
    if value in INVALID_VALUES:
        return pd.NaT

    # suppression espaces parasites
    value = value.replace(" ", "")

    parts = value.split(":")

    # format MM:SS
    if len(parts) == 2:
        mm, ss = parts
        hh = "00"

    # format HH:MM:SS
    elif len(parts) == 3:
        hh, mm, ss = parts

    else:
        return pd.NaT

    # padding
    hh = hh.zfill(2)
    mm = mm.zfill(2)
    ss = ss.zfill(2)

    normalized = f"{hh}:{mm}:{ss}"

    try:
        return pd.to_timedelta(normalized)
    except:
        return pd.NaT
```

```python
df["temps_clean"] = df["Temps"].apply(normalize_time)
```

```python
df = df[df["temps_clean"].notna()]
```

```python
df["temps_sec"] = df["temps_clean"].dt.total_seconds().astype("int32")
```
