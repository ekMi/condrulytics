```python
# Coureurs ayant plusieurs valeurs de sexe différentes
incoherent_sexe = (
    df.groupby("Nom_clean")["Sexe"]
    .nunique()
    .reset_index()
)

# filtrer ceux qui ont plus d'un sexe
incoherent_sexe = incoherent_sexe[incoherent_sexe["Sexe"] > 1]

# récupérer la liste des noms
liste_coureurs_incoherents = incoherent_sexe["Nom_clean"].tolist()
```

```python
df[df["Nom_clean"].isin(liste_coureurs_incoherents)] \
    .sort_values(["Nom_clean", "Date"])
```
